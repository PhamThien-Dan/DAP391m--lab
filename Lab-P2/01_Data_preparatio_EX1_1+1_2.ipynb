```python
import pandas as pd
titanic = pd.read_csv('titanic.csv')

print(titanic.sex.unique())
print(titanic.sex.value_counts())
SexDict = {0: 'Male', 1: 'Female'}
titanic["sex"]=titanic.sex.map(SexDict).astype('category')


print(titanic.embarked.unique())
print(titanic.embarked.value_counts())
EmbDict = {'C': 'Cherbourg', 'Q': 'Queenstown', 'S': 'Southampton'}
titanic["embarked"]=titanic.embarked.map(EmbDict).astype('category')


print(titanic.survived.unique())
print(titanic.survived.value_counts())
SurDict = {0: 'No', 1: 'Yes'}
titanic["survived"]=titanic.survived.map(SurDict).astype('category')


print(titanic.pclass.unique())
print(titanic.pclass.value_counts())
PclDict = {1: "1st", 2: "2nd",
       3: "3rd"}
titanic["pclass"]=titanic.pclass.map(PclDict)
titanic['pclass'] = pd.Categorical(titanic['pclass'], categories=['1st', '2nd', '3rd'], ordered=True)

titanic.info()
titanic[['pclass', 'survived', 'sex', 'embarked']].head()
```
