```python
# Chefboost is available at: https://github.com/serengil/chefboost

from chefboost import Chefboost as cheft
import pandas as pd
df = pd.read_csv("/content/golf.txt")
df.head()
```

```python
print("ID3 algorithm\n")

#set algorithm
config = {'algorithm': 'ID3'}


#train model on training dataset
print("Train model on traing dataset:")
model = cheft.fit(df[:8].copy(), config)

#define x vector for given day
xrow = ['Sunny','Hot','High','Weak']

#make prediction for given day
prediction = cheft.predict(model, xrow)
print("Prediction for given day. Should we play? {0}".format(prediction))

c=0
print("\nCompare predictions and expected results on TESTING dataset:")
#prediction on testing data set
for index, instance in df[8:].iterrows():
    prediction = cheft.predict(model, instance)
    actual=instance['Decision']
    print(actual,"-",prediction)
    if(actual==prediction):
        c=c+1
print("Accuracy on TESTING data set:{0}%".format(round(c/df[8:].shape[0],4) ))
```

```python
print("C4.5 algorithm\n")

#set algorithm
config = {'algorithm': 'C4.5'}
model = cheft.fit(df[:8], config)

#define xrow
xrow  = ['Sunny','Hot','High','Weak']

#make single prediction
prediction = cheft.predict(model,xrow)

#prediction on test data set
for index, instance in df[8:].iterrows():
    prediction = cheft.predict(model, instance)
    actual=instance['Decision']

    print(actual,"-",prediction)
print("Accuracy on TESTING data set:{0}%".format(round(c/df[8:].shape[0],4) ))
```

```python
print("C5.0 algorithm (Simulated using C4.5 + Adaboost)\n")

# Reload the golf.txt dataset for this example
df = pd.read_csv("/content/golf.txt")

# Create a copy of the dataframe for C5.0 specific preprocessing
df_c50_processed = df.copy()

# Identify categorical features and target
categorical_cols = ['Outlook', 'Temp.', 'Humidity', 'Wind']
target_col = 'Decision'

# Perform one-hot encoding on categorical features
df_c50_processed = pd.get_dummies(df_c50_processed, columns=categorical_cols)

# Map the target variable to numerical
df_c50_processed[target_col] = df_c50_processed[target_col].map({'Yes': 1, 'No': 0})

# 1. Cấu hình C5.0 (C4.5 + Boosting)
config = {
    'algorithm': 'C4.5',
    'enableAdaboost': True,
    'num_of_weak_classifier': 5
}

# 2. Train model trên tập dữ liệu Golf (8 dòng đầu) using the processed dataframe
model = cheft.fit(df_c50_processed[:8].copy(), config)

# 3. Define xrow (Dữ liệu lẻ để test giống các bài trên)
# For single prediction, create a dummy DataFrame and one-hot encode it,
# ensuring it has all the same columns as the training data, filled with 0s if not present.
single_prediction_data = pd.DataFrame([['Sunny','Hot','High','Weak']], columns=categorical_cols)

# Get all feature columns from the processed dataframe (excluding target)
feature_cols_processed = [col for col in df_c50_processed.columns if col != target_col]

# One-hot encode the single_prediction_data, aligning columns with df_c50_processed
single_prediction_encoded = pd.get_dummies(single_prediction_data, columns=categorical_cols)

# Add any missing columns that were in the training features and fill with 0
for col in feature_cols_processed:
    if col not in single_prediction_encoded.columns:
        single_prediction_encoded[col] = 0
# Ensure the columns are in the same order as the training features
single_prediction_encoded = single_prediction_encoded[feature_cols_processed]

xrow_for_prediction = single_prediction_encoded.iloc[0].tolist()

# 4. Make single prediction
prediction = cheft.predict(model, xrow_for_prediction)
# print("Prediction for given day:", prediction)

# 5. Prediction on test data set
print("\nCompare predictions and expected results on TESTING dataset:")

c = 0 # Biến đếm số lần đúng

for index, instance in df_c50_processed[8:].iterrows(): # Use preprocessed df for test set
    # Extract features and actual target from the preprocessed instance
    features_instance = instance[feature_cols_processed].tolist()
    actual = instance[target_col]

    prediction = cheft.predict(model, features_instance)

    print(f"{actual} - {prediction}")

    # Kiểm tra đúng sai
    if str(actual) == str(prediction):
        c += 1

# 6. Tính độ chính xác
# df_c50_processed[8:].shape[0] là số lượng dòng test
accuracy = round(c / df_c50_processed[8:].shape[0], 4) * 100
print("Accuracy on TESTING data set: {0}%".format(accuracy))
```

#Example3 (with scikit-learn.sklearn)

```python
#implementation details at https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html

from sklearn.datasets import load_iris
from sklearn import tree
X, y = load_iris(return_X_y=True)
tSize=int(X.shape[0]*0.9)
print(tSize)
X_train=X[0:tSize,:]
y_train=y[0:tSize]

X_test=X[tSize:,:]
y_test=y[tSize:]

clf = tree.DecisionTreeClassifier()

#train decision tree on training set
clf = clf.fit(X_train, y_train)

#predict values on testing set
y_pred=clf.predict(X_test);



tree.plot_tree(clf)
#!pip install graphviz

from sklearn.metrics import classification_report, confusion_matrix

print("Confusion matrix results:")
print(confusion_matrix(y_test,y_pred))

print("Classification report")
print(classification_report(y_test, y_pred))
```
